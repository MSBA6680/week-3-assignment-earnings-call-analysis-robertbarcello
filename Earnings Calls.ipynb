```python
import os
import fitz  # PyMuPDF
from textblob import TextBlob
from collections import Counter
import re

# Define the path to your folder
folder_path = r"C:\Users\rober\OneDrive\Desktop\Financial Analytics\Earnings Calls"

# Function to extract text from a PDF file
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = ""
    for page in doc:
        text += page.get_text()
    return text

# Function to perform basic text analysis
def analyze_text(text):
    blob = TextBlob(text)
    words = re.findall(r'\b\w+\b', text.lower())
    word_count = len(words)
    sentiment = blob.sentiment
    most_common_words = Counter(words).most_common(10)

    return {
        "word_count": word_count,
        "polarity": sentiment.polarity,
        "subjectivity": sentiment.subjectivity,
        "most_common_words": most_common_words
    }

# Loop through all PDF files in the folder
results = {}
for filename in os.listdir(folder_path):
    if filename.endswith(".pdf"):
        file_path = os.path.join(folder_path, filename)
        text = extract_text_from_pdf(file_path)
        analysis = analyze_text(text)
        results[filename] = analysis

# Display results
for filename, metrics in results.items():
    print(f"\n📄 {filename}")
    print(f"   - Word Count: {metrics['word_count']}")
    print(f"   - Polarity: {metrics['polarity']:.3f}")
    print(f"   - Subjectivity: {metrics['subjectivity']:.3f}")
    print(f"   - Most Common Words: {metrics['most_common_words']}")
```

```python
import os
import PyPDF2
import nltk
import gensim
from gensim import corpora
from nltk.corpus import stopwords
from nltk.tokenize import TreebankWordTokenizer
from nltk import FreqDist
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

# Ensure required NLTK resources are downloaded
try:
    nltk.data.find('sentiment/vader_lexicon')
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('vader_lexicon')
    nltk.download('stopwords')

# Initialize tokenizer
tokenizer = TreebankWordTokenizer()

def extract_text_from_pdf(pdf_file_path):
    """Extracts text from a PDF file."""
    text = ""
    with open(pdf_file_path, "rb") as file:
        reader = PyPDF2.PdfReader(file)
        for page in reader.pages:
            text += page.extract_text() or ""
    return text

def extract_sections(text):
    """Extracts Management Discussion and Q&A sections from the text."""
    management_discussion = ""
    question_answer = ""
    lines = text.splitlines()
    in_management_discussion = False
    in_question_answer = False

    for line in lines:
        line = line.strip()
        upper_line = line.upper()

        if any(phrase in upper_line for phrase in ["MANAGEMENT DISCUSSION", "MANAGEMENT'S DISCUSSION"]):
            in_management_discussion = True
            in_question_answer = False
            continue
        elif any(phrase in upper_line for phrase in ["QUESTION AND ANSWER", "Q&A", "QUESTIONS AND ANSWERS"]):
            in_question_answer = True
            in_management_discussion = False
            continue

        if in_management_discussion:
            management_discussion += line + " "
        elif in_question_answer:
            question_answer += line + " "

    return management_discussion.strip(), question_answer.strip()

def preprocess_text(text):
    """Preprocess the text for analysis."""
    tokens = tokenizer.tokenize(text)
    tokens = [word.lower() for word in tokens if word.isalpha()]
    stop_words = set(stopwords.words('english'))
    return [word for word in tokens if word not in stop_words]

def analyze_sentiment(tokens):
    """Analyze sentiment of the tokens and return positive/negative words and score."""
    sia = SentimentIntensityAnalyzer()
    positive_words = []
    negative_words = []
    compound_scores = []

    for word in tokens:
        score = sia.polarity_scores(word)['compound']
        compound_scores.append(score)
        if score > 0.05:
            positive_words.append(word)
        elif score < -0.05:
            negative_words.append(word)

    avg_score = sum(compound_scores) / len(compound_scores) if compound_scores else 0
    return positive_words, negative_words, avg_score

def perform_topic_modeling(documents, num_topics=3):
    """Performs LDA topic modeling on the given list of documents."""
    dictionary = corpora.Dictionary(documents)
    corpus = [dictionary.doc2bow(doc) for doc in documents]
    lda_model = gensim.models.LdaModel(corpus, num_topics=num_topics, id2word=dictionary, passes=10)
    return lda_model

def report_topics(lda_model, num_words=5):
    """Prints the top words in each topic."""
    for idx, topic in lda_model.print_topics(num_words=num_words):
        print(f"🧠 Topic {idx + 1}: {topic}")

def plot_topics(lda_model, num_words=5):
    """Plots top words by weight for each topic."""
    for idx, topic in lda_model.print_topics(num_words=num_words):
        words = [w.split('*')[1].strip('"') for w in topic.split(' + ')]
        weights = [float(w.split('*')[0]) for w in topic.split(' + ')]
        plt.figure(figsize=(10, 5))
        sns.barplot(x=weights, y=words)
        plt.title(f'Top Words in Topic {idx + 1}')
        plt.xlabel('Weight')
        plt.ylabel('Words')
        plt.show()

def plot_wordcloud(words, title):
    """Plots a word cloud for the given list of words."""
    if words:
        wordcloud = WordCloud(width=800, height=400, background_color='white').generate(' '.join(words))
        plt.figure(figsize=(10, 5))
        plt.imshow(wordcloud, interpolation='bilinear')
        plt.axis('off')
        plt.title(title)
        plt.show()

def analyze_section(text, section_name):
    """Performs full sentiment and topic analysis for a text section."""
    if not text.strip():
        print(f"⚠️ No content found in {section_name} section.")
        return [], [], []

    tokens = preprocess_text(text)
    if not tokens:
        print(f"⚠️ No valid tokens found in {section_name} section after preprocessing.")
        return [], [], []

    pos_words, neg_words, avg_score = analyze_sentiment(tokens)

    print(f"\n🔍 Most common words in {section_name}:")
    freq_dist = FreqDist(tokens)
    print(freq_dist.most_common(10))

    print(f"👍 Positive words: {set(pos_words)}")
    print(f"👎 Negative words: {set(neg_words)}")
    print(f"📊 Average Sentiment Score: {avg_score:.4f}")

    return tokens, pos_words, neg_words

def main():
    pdf_directory = r"C:\Users\rober\OneDrive\Desktop\Financial Analytics\Earnings Calls"

    for filename in os.listdir(pdf_directory):
        if filename.endswith(".pdf"):
            pdf_file_path = os.path.join(pdf_directory, filename)
            print(f"\n==============================")
            print(f"📄 Analyzing: {filename}")
            print(f"==============================")

            text = extract_text_from_pdf(pdf_file_path)
            mgmt_text, qa_text = extract_sections(text)

            # --- Management Discussion ---
            print(f"\n--- Management Discussion ---")
            md_tokens, md_pos, md_neg = analyze_section(mgmt_text, "Management Discussion")
            if md_tokens:
                md_lda = perform_topic_modeling([md_tokens])
                report_topics(md_lda)
                plot_topics(md_lda)
                plot_wordcloud(md_pos, "Positive Words - Management Discussion")
                plot_wordcloud(md_neg, "Negative Words - Management Discussion")

            # --- Question and Answer ---
            print(f"\n--- Question & Answer ---")
            qa_tokens, qa_pos, qa_neg = analyze_section(qa_text, "Q&A")
            if qa_tokens:
                qa_lda = perform_topic_modeling([qa_tokens])
                report_topics(qa_lda)
                plot_topics(qa_lda)
                plot_wordcloud(qa_pos, "Positive Words - Q&A")
                plot_wordcloud(qa_neg, "Negative Words - Q&A")

if __name__ == "__main__":
    main()
```

```python
import os
import PyPDF2
import nltk
from nltk.tokenize.punkt import PunktSentenceTokenizer, PunktTrainer
import pandas as pd
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, pipeline

# === NLTK CONFIGURATION ===
# Set NLTK data path explicitly
nltk_data_path = r'C:\Users\rober\nltk_data'
nltk.data.path.append(nltk_data_path)

# Try downloading punkt
nltk.download('punkt', download_dir=nltk_data_path)

# Load tokenizer manually to avoid punkt_tab bug
with open(nltk.data.find("tokenizers/punkt/english.pickle"), "rb") as f:
    sentence_tokenizer = nltk.data.pickle.load(f)

# === FINBERT SETUP ===
tokenizer = AutoTokenizer.from_pretrained("ProsusAI/finbert")
sentiment_pipeline = pipeline("text-classification", model="ProsusAI/finbert")

# Folder containing earnings call PDFs
folder_path = r"C:\Users\rober\OneDrive\Desktop\Financial Analytics\Earnings Calls"

# === PDF TEXT EXTRACTION ===
def extract_text_from_pdf(pdf_path):
    text = ""
    with open(pdf_path, "rb") as pdf_file:
        reader = PyPDF2.PdfReader(pdf_file)
        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    return text

# === SENTIMENT ANALYSIS FUNCTION ===
def analyze_text_with_truncation(text):
    truncated_text = text[:512]
    sentiment = sentiment_pipeline(truncated_text)[0]
    return sentiment

# === MAIN PROCESSING LOOP ===
sentiment_results = []

for filename in os.listdir(folder_path):
    if filename.endswith(".pdf"):
        file_path = os.path.join(folder_path, filename)
        transcript_text = extract_text_from_pdf(file_path)

        if not transcript_text.strip():
            print(f"Warning: No text found in {filename}")
            continue

        # Use loaded tokenizer to avoid the LookupError
        sentences = sentence_tokenizer.tokenize(transcript_text)

        sentiments = [analyze_text_with_truncation(sentence) for sentence in sentences]

        sentiment_counts = {
            "positive": sum(1 for s in sentiments if s["label"] == "positive"),
            "neutral": sum(1 for s in sentiments if s["label"] == "neutral"),
            "negative": sum(1 for s in sentiments if s["label"] == "negative")
        }

        sentiment_results.append({
            "Earnings Call": filename,
            "Positive": sentiment_counts["positive"],
            "Neutral": sentiment_counts["neutral"],
            "Negative": sentiment_counts["negative"],
            "Total Sentences": len(sentiments),
            "Positive %": sentiment_counts["positive"] / len(sentiments) * 100,
            "Negative %": sentiment_counts["negative"] / len(sentiments) * 100
        })

# === CREATE RESULTS DATAFRAME ===
df_results = pd.DataFrame(sentiment_results)

# === PLOT SENTIMENT ===
plt.figure(figsize=(10, 5))
plt.plot(df_results["Earnings Call"], df_results["Positive %"], marker="o", linestyle="-", label="Positive Sentiment (%)")
plt.plot(df_results["Earnings Call"], df_results["Negative %"], marker="s", linestyle="--", label="Negative Sentiment (%)")
plt.xlabel("Earnings Call")
plt.ylabel("Sentiment Percentage")
plt.title("Sensitivity Analysis of Earnings Call Sentiments (FinBERT)")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# === PRINT RESULTS ===
print(df_results)
```
